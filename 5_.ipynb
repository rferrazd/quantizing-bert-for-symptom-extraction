```python
from datasets import load_dataset, Features, Sequence, Value

features = Features({
      "text": Value("string"),
      "word_tokens": Sequence(Value("string")),
      "labels": Sequence(Value("string")),
      "tokens": Sequence(Value("string")),
      "input_ids": Sequence(Value("int32")),
      "token_labels": Sequence(Value("string")),  # or make all int32 if you drop the "O"/"B-" strings
      "token_label_ids": Sequence(Value("int32")),
  })

dataset = load_dataset("json", data_files=data_files, features=features)
```
